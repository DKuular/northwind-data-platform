# Spark quickstart (northwind-jupyter)

В compose уже заданы `PYTHONPATH` → `modules` и `SPARK_CONFIG_PATH` → `config.json`. Пока ячейки со Spark выполняются, UI драйвера доступен на хосте: **http://localhost:4040** (если проброшен порт `4040:4040`). После `spark.stop()` страница пропадает.

Ниже используется **`local[*]`**, чтобы не зависеть от очереди на Spark Standalone.

```python
import spark_config
from pyspark.sql import SparkSession

cfg = spark_config.load_config()

spark = (
    SparkSession.builder.appName("jupyter-quickstart")
    .master("local[*]")
    .config("spark.sql.shuffle.partitions", cfg["spark"]["shuffle_partitions"])
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print("Master в конфиге (для spark-submit):", cfg["spark"]["master"])
```

```python
spark.range(100).selectExpr("id", "cast(id % 10 as int) as bucket").groupBy("bucket").count().orderBy("bucket").show()
```
